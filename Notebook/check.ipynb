```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------------
# Target-wise plot (uses existing plot_df)
# -----------------------------------

target_order = ["LD2", "C", "Cu", "Sn"]
target_order = [t for t in target_order if t in plot_df["target"].unique()]

# unique Q2 bins
q2_bins = (
    plot_df[["Q2_min", "Q2_max"]]
    .drop_duplicates()
    .sort_values(["Q2_min", "Q2_max"])
    .values.tolist()
)

# styles for different Q2 bins
q2_labels = [f"{q2min:.1f} < Q² < {q2max:.1f}" for q2min, q2max in q2_bins]
colors = ["black", "red", "blue", "green", "purple", "orange", "brown", "magenta"]
markers = ["o", "s", "^", "D", "v", "P", "X", "*"]

style_map = {}
for i, ((q2min, q2max), label) in enumerate(zip(q2_bins, q2_labels)):
    style_map[(q2min, q2max)] = {
        "label": label,
        "color": colors[i % len(colors)],
        "marker": markers[i % len(markers)]
    }

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)
axes = axes.flatten()

# show tick labels on all panels
for ax in axes:
    ax.tick_params(axis='x', labelbottom=True, labelsize=12)
    ax.tick_params(axis='y', labelleft=True, labelsize=12)

x_min = plot_df["xB_center"].min()
x_max = plot_df["xB_center"].max()
y_min = plot_df["ratio"].min()
y_max = plot_df["ratio"].max()

x_pad = 0.02 * (x_max - x_min)
y_pad = 0.05 * (y_max - y_min)

for ax, tgt in zip(axes, target_order):
    sub_t = plot_df[plot_df["target"] == tgt].copy()

    for (q2min, q2max) in q2_bins:
        ss = sub_t[
            (sub_t["Q2_min"] == q2min) &
            (sub_t["Q2_max"] == q2max)
        ].sort_values("xB_center")

        if len(ss) == 0:
            continue

        st = style_map[(q2min, q2max)]

        ax.plot(
            ss["xB_center"],
            ss["ratio"],
            marker=st["marker"],
            color=st["color"],
            linestyle="-",
            linewidth=1.2,
            markersize=6,
            label=st["label"]
        )

    ax.set_title(f"{tgt}", fontsize=16)
    ax.set_xlim(x_min - x_pad, x_max + x_pad)
    ax.set_ylim(y_min - y_pad, y_max + y_pad)
    ax.set_xlabel(r"$x_B$", fontsize=16)
    ax.set_ylabel(r"$\sigma_{\mathrm{rad}}/\sigma_{\mathrm{born}}$", fontsize=16)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=10)

# turn off unused panels if fewer than 4 targets
for j in range(len(target_order), len(axes)):
    axes[j].axis("off")

# increase spacing between subplots
plt.subplots_adjust(wspace=0.28, hspace=0.32)

plt.show()
```
