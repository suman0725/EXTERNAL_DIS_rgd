```python
import matplotlib.pyplot as plt

elements = [
    ("Al entrance\n30 μm", -8.0),
    ("Al exit\n30 μm", -3.0),
    ("MLI barrier\n15 μm", 0.0),
    ("SC exit\n75 μm", 22.3),
]

ld2_start = -7.92
ld2_end   = -3.05

fig, ax = plt.subplots(figsize=(10, 2.5))

# beamline
ax.hlines(0, -20, 25, linewidth=2)

# LD2 blue region
ax.axvspan(ld2_start, ld2_end, alpha=0.3)
ax.text((ld2_start + ld2_end)/2, -0.28, "LD2 cell", ha="center", va="top", fontsize=10)

# markers and labels
for label, z in elements:
    ax.vlines(z, -0.18, 0.18, linewidth=2)
    ax.text(z, 0.25, label, ha="center", va="bottom", fontsize=9)

ax.set_xlim(-25, 25)
ax.set_xticks([-20, -15, -10, -8, -5, -3, 0, 5, 10, 15, 20, 25])  # keep old ticks and add -8, -3
ax.set_ylim(-0.5, 0.8)
ax.set_xlabel("z [cm]")
ax.set_yticks([])
ax.set_title("")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

elements = [
    ("Cu foil\n93 μm", -8.0),
    ("Sn foil\n171 μm", -3.0),
]

ld2_start = -17.92
ld2_end   = -13.05

fig, ax = plt.subplots(figsize=(10, 2.5))

# beamline
ax.hlines(0, -20, 25, linewidth=2)

# empty LD2 cell as blue boundary only
ax.axvspan(ld2_start, ld2_end, facecolor='none', edgecolor='blue', linewidth=2)
ax.text((ld2_start + ld2_end)/2, -0.28, "empty LD2 cell", ha="center", va="top", fontsize=10, color="blue")

# foil markers and labels
for label, z in elements:
    ax.vlines(z, -0.18, 0.18, linewidth=2)
    ax.text(z, 0.25, label, ha="center", va="bottom", fontsize=9)

ax.set_xlim(-20, 25)
ax.set_xticks([-20, -18, -15, -13, -10, -8, -5, -3, 0, 5, 10, 15, 20, 25])
ax.set_ylim(-0.5, 0.8)
ax.set_xlabel("z [cm]")
ax.set_yticks([])
ax.set_title("")

plt.tight_layout()
plt.show()
```
